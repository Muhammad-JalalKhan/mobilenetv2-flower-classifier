```python
import torch, random, numpy as np, os
from torchvision import datasets, transforms
from torch.utils.data import Subset, DataLoader

SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)

# fixed 45,000 / 5,000 split of the 50,000 training images
g = torch.Generator().manual_seed(SEED)
perm = torch.randperm(50000, generator=g).tolist()
train_idx, val_idx = perm[:45000], perm[45000:]

# raw images, used only to compute statistics
raw = datasets.CIFAR10("./data", train=True, download=True, transform=transforms.ToTensor())
loader = DataLoader(Subset(raw, train_idx), batch_size=500)

n, s, s2 = 0, torch.zeros(3), torch.zeros(3)
for x, _ in loader:
    n += x.size(0) * x.size(2) * x.size(3)
    s += x.sum(dim=[0, 2, 3])
    s2 += (x ** 2).sum(dim=[0, 2, 3])
mean = s / n
std = (s2 / n - mean ** 2).sqrt()
print("Train-only mean:", mean)
print("Train-only std: ", std)
```

```python
M, S = mean.tolist(), std.tolist()
tf_train = transforms.Compose([transforms.Resize(128), transforms.RandomHorizontalFlip(),
                               transforms.ToTensor(), transforms.Normalize(M, S)])
tf_eval  = transforms.Compose([transforms.Resize(128), transforms.ToTensor(),
                               transforms.Normalize(M, S)])

train_full = datasets.CIFAR10("./data", train=True, transform=tf_train)
eval_full  = datasets.CIFAR10("./data", train=True, transform=tf_eval)
train_set = Subset(train_full, train_idx)    # augmentation on
val_set   = Subset(eval_full, val_idx)       # no augmentation
test_set  = datasets.CIFAR10("./data", train=False, transform=tf_eval)

print(len(train_set), len(val_set), len(test_set))   # 45000 5000 10000

train_loader = DataLoader(train_set, batch_size=32, shuffle=True)
x, y = next(iter(train_loader))
print(x.shape, x.dtype)    # [32, 3, 128, 128] torch.float32
```

```python
import time
from torch.amp import autocast, GradScaler
import torch.nn as nn

device = "cuda"
CKPT_DIR = "/content/drive/MyDrive/mbv2_checkpoints"
os.makedirs(CKPT_DIR, exist_ok=True)

val_loader  = DataLoader(val_set,  batch_size=256, num_workers=2, pin_memory=True)
test_loader = DataLoader(test_set, batch_size=256, num_workers=2, pin_memory=True)

def evaluate(model, loader):
    model.eval(); correct = total = 0
    with torch.no_grad(), autocast("cuda"):
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            correct += (model(xb).argmax(1) == yb).sum().item()
            total += yb.size(0)
    return 100 * correct / total

def build_model(alpha):
    # TEMPORARY stand-in. Replace with Jalal's class from src/mobilenetv2.py
    from torchvision.models import mobilenet_v2
    return mobilenet_v2(width_mult=alpha, num_classes=10, dropout=0.2)
```

```python
def train_width(alpha, name, epochs=15, lr=1e-3, bs=128):
    torch.manual_seed(SEED)
    model = build_model(alpha).to(device)
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-4)
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs)
    scaler = GradScaler("cuda")
    loss_fn = nn.CrossEntropyLoss()
    loader = DataLoader(train_set, batch_size=bs, shuffle=True, num_workers=2, pin_memory=True)

    last_path = f"{CKPT_DIR}/{name}_last.pt"
    best_path = f"{CKPT_DIR}/{name}_best.pt"
    start, best_val = 0, 0.0
    if os.path.exists(last_path):                      # resume after a disconnect
        ck = torch.load(last_path, map_location=device)
        model.load_state_dict(ck["model"]); opt.load_state_dict(ck["opt"])
        sched.load_state_dict(ck["sched"]); scaler.load_state_dict(ck["scaler"])
        start, best_val = ck["epoch"] + 1, ck["best_val"]
        print(f"Resuming from epoch {start}")

    for ep in range(start, epochs):
        model.train(); t0 = time.time(); running = 0
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            opt.zero_grad()
            with autocast("cuda"):
                loss = loss_fn(model(xb), yb)
            scaler.scale(loss).backward()
            scaler.step(opt); scaler.update()
            running += loss.item()
        sched.step()
        val = evaluate(model, val_loader)
        if val > best_val:
            best_val = val
            torch.save(model.state_dict(), best_path)
        torch.save({"model": model.state_dict(), "opt": opt.state_dict(),
                    "sched": sched.state_dict(), "scaler": scaler.state_dict(),
                    "epoch": ep, "best_val": best_val}, last_path)
        print(f"ep {ep+1}/{epochs}  loss {running/len(loader):.3f}  val {val:.2f}  best {best_val:.2f}  {time.time()-t0:.0f}s")

    model.load_state_dict(torch.load(best_path, map_location=device))
    return model, best_val
```

```python
import copy, subprocess, pandas as pd
from thop import profile

CPU_NAME = subprocess.getoutput("lscpu | grep 'Model name'").split(":")[-1].strip()
print("CPU:", CPU_NAME)

def measure(model, res=128):
    m = copy.deepcopy(model).cpu().eval()
    params_m = sum(p.numel() for p in m.parameters()) / 1e6
    macs, _ = profile(m, inputs=(torch.randn(1, 3, res, res),), verbose=False)
    xb = torch.randn(1, 3, res, res)
    with torch.no_grad():
        for _ in range(20): m(xb)
        t0 = time.perf_counter()
        for _ in range(100): m(xb)
        lat_ms = (time.perf_counter() - t0) / 100 * 1000
    return params_m, macs / 1e6, lat_ms

print(measure(model))
```

```python
TAG = "_standin"    # change to "" when you switch to Jalal's model
RESULTS_DRIVE = f"{CKPT_DIR}/width_sweep_results{TAG}.csv"
widths = [(0.35, "W01"), (0.50, "W02"), (0.75, "W03"), (1.00, "W04")]

rows = pd.read_csv(RESULTS_DRIVE).to_dict("records") if os.path.exists(RESULTS_DRIVE) else []
done = {r["Run ID"] for r in rows}

for alpha, rid in widths:
    if rid in done:
        print("skipping", rid, "(already finished)"); continue
    name = f"mbv2_w{int(round(alpha*100)):03d}_r128{TAG}"
    print(f"\n=== {rid}: alpha={alpha} ===")
    model, best_val = train_width(alpha, name, epochs=20)
    test_acc = evaluate(model, test_loader)
    p, macs, lat = measure(model)
    rows.append({"Run ID": rid, "Width (α)": alpha, "Input Res": "128×128",
                 "Params (M)": round(p, 2), "MACs (M)": round(macs, 1),
                 "CPU Latency (ms, BS=1)": round(lat, 1),
                 "Top-1 Val Acc (%)": round(best_val, 2),
                 "Top-1 Acc (%)": round(test_acc, 2),
                 "Drive Checkpoint File": f"{name}_best.pt",
                 "CPU": CPU_NAME})
    df = pd.DataFrame(rows)
    df.to_csv(RESULTS_DRIVE, index=False)
    os.makedirs("results", exist_ok=True)
    df.to_csv("results/width_sweep_results.csv", index=False)
    print(rows[-1])
```
